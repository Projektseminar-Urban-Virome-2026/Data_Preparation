# Imports

```python
from pathlib import Path # Importing the Path class from the pathlib module to handle file paths
import pandas as pd # Importing the pandas library for data manipulation and analysis
```

# Data Import

```python
# Define the path to the CSV file containing virus information
csv_file = Path("../viruses_cleaned.csv")

# Function to load virus information from the specified CSV file
def load_virus_information(source):
    return pd.read_csv(source, sep=",")

# Load the virus information into a DataFrame
df = load_virus_information(csv_file)

# Checking the DataFrame to ensure it contains the expected columns and data
print("Total number of samples in the dataset:", len(df))
print("\n================================\n")

print(df.head())  # Shows the first five rows when the script is executed.
```

# Filter for Humans

```python
# Removing all rows where the 'host tax id' is not 9606 (which corresponds to humans as hosts)
df_reduced = df[df['host tax id'] == 9606]

# Display the first rows of the updated DataFrame to verify the removal of the 'host tax id' and 'host name' columns
print(df_reduced.head())  # Shows the first five rows of the updated DataFrame when the script is executed.
```

# Save Data As New CSV file

```python
# Saving the cleaned DataFrame to a new CSV file
output_csv_file = Path("../generated_files/viruses_cleaned_human.csv")
df_reduced.to_csv(output_csv_file, index=False)
```
